```python
# Cell 3: Real-Time SMT Logits Masking Operator Demonstration
from model.logits_operator import SMTLogitsOperator

operator = SMTLogitsOperator(classification_level="CUI")
candidate_tokens = ["UNRESTRICTED_LOG", "RESTRICTED_CUI_DATA", "UNENCRYPTED_PUBLIC_EMISSION"]
raw_logits = [2.5, 9.1, 8.8]

masked_logits = operator.apply_logits_mask(candidate_tokens, raw_logits)

print("--- SMT CONSTRAINED DECODING RESULTS ---")
for tok, raw, masked in zip(candidate_tokens, raw_logits, masked_logits):
    print(f"Token: {tok:<30} | Raw Logit: {raw:<5.1f} | Masked Logit: {masked}")
```
